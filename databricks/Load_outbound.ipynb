#####This notebook will be used to load the outbound layer of the warehouse. The layer will then be used to feed data to snowflake tables.

```python
import boto3
import datetime
```

```python
EXTRACT_PATH = "s3a://nyc-lakehouse/outbound/"
bucket = "nyc-lakehouse"
```

```python
# Initialize client
ACCESS_KEY = dbutils.secrets.get(scope="my_app_creds", key="access_key_id")
SECRET_KEY = dbutils.secrets.get(scope="my_app_creds", key="secret_access_key")
REGION = 'us-east-1'
```

```python
date_today = datetime.datetime.now().strftime("%Y-%m-%d")
```

```python
# Create a client with explicit credentials
s3 = boto3.client(
    's3',
    aws_access_key_id=ACCESS_KEY,
    aws_secret_access_key=SECRET_KEY,
    region_name=REGION
)

```

```python
prefix_daily = f"outbound/outbound_daily_stats/"
response_daily = s3.list_objects_v2(Bucket=bucket, Prefix=prefix_daily)

for obj in response_daily.get("Contents", []):
    s3.delete_object(Bucket=bucket, Key=obj["Key"])

(
spark.read.format("delta")
    .table("nyc_gold_daily_stats")
    .write
    .mode("overwrite")
    .parquet(f"s3a://nyc-lakehouse/outbound/outbound_daily_stats/{date_today}")
)
```
