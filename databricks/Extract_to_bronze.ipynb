```python
dbutils.widgets.text("date_code_param", "")
dbutils.widgets.text("dest_folder", defaultValue= "")
```

```python
import boto3
import requests
```

```python

ACCESS_KEY = dbutils.secrets.get(scope="my_app_creds", key="access_key_id")
SECRET_KEY = dbutils.secrets.get(scope="my_app_creds", key="secret_access_key")
REGION = 'us-east-1'

# Create a client with explicit credentials
s3_client = boto3.client(
    's3',
    aws_access_key_id=ACCESS_KEY,
    aws_secret_access_key=SECRET_KEY,
    region_name=REGION
)
```

```python
#Date code should be provided in format yyyy-mm
date_code = dbutils.widgets.get("date_code_param")
print(f"Extracting for for date code {date_code}")
if not date_code:
    dbutils.notebook.exit("No date code provided")
```

```python

url = "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_" + date_code + ".parquet"
bucket = "nyc-lakehouse"
destination_folder = dbutils.widgets.get("dest_folder")
key = f"bronze/{destination_folder}yellow_tripdata_{date_code}.parquet"

with requests.get(url, stream=True) as r:
    r.raise_for_status()
    s3_client.upload_fileobj(r.raw, bucket, key)
```
